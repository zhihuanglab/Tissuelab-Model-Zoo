```python
import requests
import json

BASE_URL = "http://127.0.0.1:5001"


def test_gpt_chat():
    data = {
        "agent_id": "agent1",
        "prompt": "What is the capital of France?",
        "parameters": {}
    }
    response = requests.post(f"{BASE_URL}/api/agent/v1/chat", json=data)
    print("Chat Response:", response.status_code)
    print(json.dumps(response.json(), indent=2))


# Run test
print("Testing GPT Chat...")
test_gpt_chat()
```

```python
def test_get_steps(message):
    data = {"agent_id": "agent1", "prompt": message, "parameters": {}}
    response = requests.post(f"{BASE_URL}/api/agent/v1/get_steps", json=data)
    print("Get Steps Response:", response.status_code)
    print(json.dumps(response.json(), indent=2))
    return response.json()
```

```python
user_query = "Can you help me count the average number of podocytes in kidney glomerulus?"
get_steps = test_get_steps(user_query)
print(get_steps)
```

```python
def test_process_script(message):
    data = {"agent_id": "agent1", "prompt": message, "parameters": {}}
    response = requests.post(f"{BASE_URL}/api/agent/v1/process_script", json=data)
    print("Process Script Response:", response.status_code)
    print(json.dumps(response.json(), indent=2))
    return response.json()

user_query = "Count the number of cells in the image."
process_script = test_process_script(user_query)
print(process_script)
```

```python
def script_function(script):
    namespace = {}
    try:
        # Execute the code in the isolated namespace
        exec(script, namespace)
        # Return the function object
        return namespace['analyze_medical_image']
    except Exception as e:
        print(f"Error executing code: {e}")
        return None


func = script_function(process_script["data"])

```
